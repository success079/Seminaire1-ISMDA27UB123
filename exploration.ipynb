## Partie 2 — Résumer : les statistiques à la main

Dans `afrishop.py` : **TODO 6** à **TODO 9** (`moyenne`, `mediane`, `variance`, `ecart_type`).

## Partie 4 — Mesurer les liens : la matrice de covariance

### Construire les trois colonnes

Rien à écrire dans `afrishop.py` pour cet exercice : tout se passe ici.

```python
# Étape 1 : construisez prix, stock et ca, trois listes parallèles sur les 20 produits,
#           dans une MÊME boucle sur les produits.
# Étape 2 : vérifiez que les trois listes ont la même longueur.
# Étape 3 : affichez la moyenne de chacune.
prix, stock, ca = [], [], []
for p in produits:
    prix.append(p["prix_unitaire"])
    stock.append(float(p["stock"]))
    ca.append(ca_produit.get(p["id_produit"], 0.0))

print("Longueurs :", len(prix), len(stock), len(ca), "-> identiques :", len(prix) == len(stock) == len(ca))
print(f"Moyenne prix  : {afrishop.moyenne(prix):10,.2f}")
print(f"Moyenne stock : {afrishop.moyenne(stock):10,.2f}")
print(f"Moyenne CA    : {afrishop.moyenne(ca):10,.2f}")
```

### Calculer la covariance et la corrélation

Dans `afrishop.py` : **TODO 16** (`covariance`) et **TODO 17** (`correlation`).

```python
# Étape 3 : calculez les trois corrélations : prix/stock, prix/CA, stock/CA.
# Étape 4 : vérifiez que covariance(xs, xs) redonne variance(xs), avec abs(a - b) < 1e-9.
print(f"corr(prix, stock) = {afrishop.correlation(prix, stock):+.3f}")
print(f"corr(prix, CA)    = {afrishop.correlation(prix, ca):+.3f}")
print(f"corr(stock, CA)   = {afrishop.correlation(stock, ca):+.3f}")

for nom, col in [("prix", prix), ("stock", stock), ("ca", ca)]:
    ok = abs(afrishop.covariance(col, col) - afrishop.variance(col)) < 1e-9 * max(1.0, afrishop.variance(col))
    print(f"covariance({nom}, {nom}) == variance({nom}) : {ok}")
```

```python
verifier("covariance", "correlation")
```

### Assembler la matrice de covariance

Dans `afrishop.py` : **TODO 18** (`matrice_covariance`).

```python
# Étape 2 : appliquez-la à [prix, stock, ca] et affichez le résultat aligné, avec un en-tête de colonnes.
# Étape 3 : vérifiez que la matrice est symétrique.
# Étape 4 : vérifiez que sa diagonale contient les variances des trois colonnes.
noms = ["prix", "stock", "ca"]
colonnes = [prix, stock, ca]
cov = afrishop.matrice_covariance(colonnes)

print(f"{'':8s}" + "".join(f"{n:>16s}" for n in noms))
for nom, ligne in zip(noms, cov):
    print(f"{nom:8s}" + "".join(f"{x:16,.2f}" for x in ligne))

symetrique = all(abs(cov[i][j] - cov[j][i]) < 1e-9 * max(1.0, abs(cov[i][j]))
                 for i in range(3) for j in range(3))
print("\nSymétrique :", symetrique)

diag_ok = all(abs(cov[i][i] - afrishop.variance(col)) < 1e-9 * max(1.0, cov[i][i])
              for i, col in enumerate(colonnes))
print("Diagonale = variances :", diag_ok)
```

```python
verifier("matrice_covariance")
```

### Extraire les directions principales

Dans `afrishop.py` : **TODO 19** (`valeurs_propres_2x2`).

```python
# Étape 1 : extrayez la sous-matrice 2 x 2 correspondant à prix et stock.
# Étape 4 : vérifiez que la somme des deux valeurs propres vaut la trace, et leur produit le déterminant.
# Étape 5 : calculez la part de variance portée par la plus grande.
sous_matrice = [[cov[0][0], cov[0][1]],
                [cov[1][0], cov[1][1]]]
print("Sous-matrice prix/stock :", sous_matrice)

l1, l2 = afrishop.valeurs_propres_2x2(sous_matrice)
print(f"lambda1 = {l1:,.2f}   lambda2 = {l2:,.2f}")

trace = sous_matrice[0][0] + sous_matrice[1][1]
det = sous_matrice[0][0] * sous_matrice[1][1] - sous_matrice[0][1] * sous_matrice[1][0]
print("Somme   = trace       :", abs((l1 + l2) - trace) < 1e-9 * trace)
print("Produit = déterminant :", abs(l1 * l2 - det) < 1e-9 * abs(det))

print(f"Part de variance du premier axe : {l1 / (l1 + l2):.1%}")
```

```python
verifier("valeurs_propres_2x2")
```
